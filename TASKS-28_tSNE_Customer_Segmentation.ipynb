```python
# ============================================================
# DAY 28 - t-SNE CUSTOMER SEGMENTATION
# PCA vs t-SNE Visualization
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import warnings
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.cluster import KMeans
from sklearn.metrics import (
    silhouette_score,
    calinski_harabasz_score,
    davies_bouldin_score
)

warnings.filterwarnings("ignore")

RANDOM_STATE = 42

OUTPUT_DIR = "day28_outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

np.random.seed(RANDOM_STATE)

print("Libraries imported successfully.")


# ============================================================
# 2. CREATE CUSTOMER DATASET
# ============================================================

n_customers = 1500

customer_df = pd.DataFrame({

    "Age": np.random.normal(
        35, 10, n_customers
    ),

    "Annual_Income": np.random.normal(
        65000, 22000, n_customers
    ),

    "Purchase_Frequency": np.random.normal(
        12, 5, n_customers
    ),

    "Average_Order_Value": np.random.normal(
        90, 30, n_customers
    ),

    "Website_Visits": np.random.normal(
        55, 20, n_customers
    ),

    "Discount_Usage": np.random.normal(
        45, 20, n_customers
    ),

    "Customer_Lifetime_Value": np.random.normal(
        3500, 1200, n_customers
    ),

    "App_Usage_Hours": np.random.normal(
        15, 7, n_customers
    ),

    "Support_Requests": np.random.normal(
        5, 3, n_customers
    ),

    "Product_Reviews": np.random.normal(
        8, 4, n_customers
    )
})


# ============================================================
# 3. CLEAN / CLIP VALUES
# ============================================================

customer_df["Age"] = customer_df[
    "Age"
].clip(18, 75)

customer_df["Annual_Income"] = customer_df[
    "Annual_Income"
].clip(15000, 200000)

customer_df["Purchase_Frequency"] = customer_df[
    "Purchase_Frequency"
].clip(1, 50)

customer_df["Average_Order_Value"] = customer_df[
    "Average_Order_Value"
].clip(5, 500)

customer_df["Website_Visits"] = customer_df[
    "Website_Visits"
].clip(1, 200)

customer_df["Discount_Usage"] = customer_df[
    "Discount_Usage"
].clip(0, 100)

customer_df["Customer_Lifetime_Value"] = customer_df[
    "Customer_Lifetime_Value"
].clip(100, 15000)

customer_df["App_Usage_Hours"] = customer_df[
    "App_Usage_Hours"
].clip(0, 60)

customer_df["Support_Requests"] = customer_df[
    "Support_Requests"
].clip(0, 30)

customer_df["Product_Reviews"] = customer_df[
    "Product_Reviews"
].clip(0, 30)

customer_df = customer_df.round(2)

customer_df.insert(
    0,
    "Customer_ID",
    range(1, len(customer_df) + 1)
)

print("Dataset shape:", customer_df.shape)

display(customer_df.head())


# ============================================================
# 4. DATA EXPLORATION
# ============================================================

print("\nDataset Information:")
customer_df.info()

print("\nMissing Values:")
print(customer_df.isnull().sum())

print("\nStatistical Summary:")
display(customer_df.describe())


# ============================================================
# 5. SAVE DATASET
# ============================================================

customer_df.to_csv(
    f"{OUTPUT_DIR}/customer_dataset.csv",
    index=False
)


# ============================================================
# 6. SELECT FEATURES
# ============================================================

features = [
    "Age",
    "Annual_Income",
    "Purchase_Frequency",
    "Average_Order_Value",
    "Website_Visits",
    "Discount_Usage",
    "Customer_Lifetime_Value",
    "App_Usage_Hours",
    "Support_Requests",
    "Product_Reviews"
]

X = customer_df[features].copy()

print(
    "\nNumber of features:",
    len(features)
)


# ============================================================
# 7. FEATURE SCALING
# ============================================================

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print(
    "Feature scaling completed."
)


# ============================================================
# 8. PCA BASELINE
# ============================================================

pca = PCA(
    n_components=2,
    random_state=RANDOM_STATE
)

X_pca = pca.fit_transform(
    X_scaled
)

customer_df["PCA_1"] = X_pca[:, 0]
customer_df["PCA_2"] = X_pca[:, 1]

pca_variance = (
    pca.explained_variance_ratio_.sum()
)

print(
    "\nPCA variance retained:",
    round(pca_variance, 4)
)


# ============================================================
# 9. PCA VISUALIZATION
# ============================================================

plt.figure(figsize=(12, 8))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    alpha=0.6
)

plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")

plt.title(
    "Customer Data - PCA 2D Visualization"
)

plt.grid(True)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/pca_visualization.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 10. t-SNE
# ============================================================

print(
    "\nRunning t-SNE..."
)

tsne = TSNE(
    n_components=2,
    perplexity=30,
    learning_rate="auto",
    init="pca",
    max_iter=1000,
    random_state=RANDOM_STATE
)

X_tsne = tsne.fit_transform(
    X_scaled
)

customer_df["TSNE_1"] = X_tsne[:, 0]
customer_df["TSNE_2"] = X_tsne[:, 1]

print(
    "t-SNE completed."
)


# ============================================================
# 11. t-SNE VISUALIZATION
# ============================================================

plt.figure(figsize=(12, 8))

plt.scatter(
    X_tsne[:, 0],
    X_tsne[:, 1],
    alpha=0.6
)

plt.xlabel("t-SNE Dimension 1")
plt.ylabel("t-SNE Dimension 2")

plt.title(
    "Customer Data - t-SNE 2D Visualization"
)

plt.grid(True)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/tsne_visualization.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 12. COMPARE PCA AND t-SNE
# ============================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(16, 6)
)

axes[0].scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    alpha=0.6
)

axes[0].set_title(
    "PCA"
)

axes[0].set_xlabel(
    "PCA Component 1"
)

axes[0].set_ylabel(
    "PCA Component 2"
)

axes[0].grid(True)


axes[1].scatter(
    X_tsne[:, 0],
    X_tsne[:, 1],
    alpha=0.6
)

axes[1].set_title(
    "t-SNE"
)

axes[1].set_xlabel(
    "t-SNE Dimension 1"
)

axes[1].set_ylabel(
    "t-SNE Dimension 2"
)

axes[1].grid(True)

plt.suptitle(
    "PCA vs t-SNE"
)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/pca_vs_tsne.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 13. K-MEANS ON ORIGINAL DATA
# ============================================================

k_values = range(2, 9)

original_results = []

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=RANDOM_STATE,
        n_init=20
    )

    labels = model.fit_predict(
        X_scaled
    )

    original_results.append({

        "K": k,

        "Inertia":
            model.inertia_,

        "Silhouette":
            silhouette_score(
                X_scaled,
                labels
            )
    })


original_results_df = pd.DataFrame(
    original_results
)

print(
    "\nK-Means on Original Data:"
)

display(
    original_results_df
)


# ============================================================
# 14. K-MEANS ON PCA DATA
# ============================================================

pca_results = []

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=RANDOM_STATE,
        n_init=20
    )

    labels = model.fit_predict(
        X_pca
    )

    pca_results.append({

        "K": k,

        "Inertia":
            model.inertia_,

        "Silhouette":
            silhouette_score(
                X_pca,
                labels
            )
    })


pca_results_df = pd.DataFrame(
    pca_results
)

print(
    "\nK-Means on PCA Data:"
)

display(
    pca_results_df
)


# ============================================================
# 15. K-MEANS ON t-SNE
# ============================================================

tsne_results = []

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=RANDOM_STATE,
        n_init=20
    )

    labels = model.fit_predict(
        X_tsne
    )

    tsne_results.append({

        "K": k,

        "Inertia":
            model.inertia_,

        "Silhouette":
            silhouette_score(
                X_tsne,
                labels
            )
    })


tsne_results_df = pd.DataFrame(
    tsne_results
)

print(
    "\nK-Means on t-SNE Data:"
)

display(
    tsne_results_df
)


# ============================================================
# 16. SAVE COMPARISON RESULTS
# ============================================================

original_results_df.to_csv(
    f"{OUTPUT_DIR}/original_kmeans_results.csv",
    index=False
)

pca_results_df.to_csv(
    f"{OUTPUT_DIR}/pca_kmeans_results.csv",
    index=False
)

tsne_results_df.to_csv(
    f"{OUTPUT_DIR}/tsne_kmeans_results.csv",
    index=False
)


# ============================================================
# 17. SILHOUETTE COMPARISON
# ============================================================

plt.figure(figsize=(12, 7))

plt.plot(
    original_results_df["K"],
    original_results_df["Silhouette"],
    marker="o",
    label="Original Data"
)

plt.plot(
    pca_results_df["K"],
    pca_results_df["Silhouette"],
    marker="s",
    label="PCA"
)

plt.plot(
    tsne_results_df["K"],
    tsne_results_df["Silhouette"],
    marker="^",
    label="t-SNE"
)

plt.xlabel(
    "Number of Clusters"
)

plt.ylabel(
    "Silhouette Score"
)

plt.title(
    "Clustering Comparison"
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/clustering_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 18. SELECT K FROM ORIGINAL DATA
# ============================================================

best_k = int(
    original_results_df.loc[
        original_results_df[
            "Silhouette"
        ].idxmax(),
        "K"
    ]
)

print(
    "\nSelected K:",
    best_k
)


# ============================================================
# 19. FINAL K-MEANS MODEL
# ============================================================

final_kmeans = KMeans(
    n_clusters=best_k,
    random_state=RANDOM_STATE,
    n_init=30
)

final_labels = (
    final_kmeans.fit_predict(
        X_scaled
    )
)

customer_df[
    "Cluster"
] = final_labels


# ============================================================
# 20. FINAL CLUSTER METRICS
# ============================================================

final_silhouette = silhouette_score(
    X_scaled,
    final_labels
)

final_calinski = (
    calinski_harabasz_score(
        X_scaled,
        final_labels
    )
)

final_davies = (
    davies_bouldin_score(
        X_scaled,
        final_labels
    )
)

print(
    "\nFinal Clustering Metrics"
)

print(
    "Silhouette Score:",
    round(final_silhouette, 4)
)

print(
    "Calinski-Harabasz Score:",
    round(final_calinski, 4)
)

print(
    "Davies-Bouldin Score:",
    round(final_davies, 4)
)


# ============================================================
# 21. t-SNE CLUSTER VISUALIZATION
# ============================================================

plt.figure(figsize=(12, 8))

scatter = plt.scatter(
    customer_df["TSNE_1"],
    customer_df["TSNE_2"],
    c=customer_df["Cluster"],
    alpha=0.7
)

plt.xlabel(
    "t-SNE Dimension 1"
)

plt.ylabel(
    "t-SNE Dimension 2"
)

plt.title(
    "Customer Clusters Visualized using t-SNE"
)

plt.colorbar(
    scatter,
    label="Cluster"
)

plt.grid(True)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/tsne_cluster_visualization.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 22. PCA CLUSTER VISUALIZATION
# ============================================================

plt.figure(figsize=(12, 8))

scatter = plt.scatter(
    customer_df["PCA_1"],
    customer_df["PCA_2"],
    c=customer_df["Cluster"],
    alpha=0.7
)

plt.xlabel(
    "PCA Component 1"
)

plt.ylabel(
    "PCA Component 2"
)

plt.title(
    "Customer Clusters Visualized using PCA"
)

plt.colorbar(
    scatter,
    label="Cluster"
)

plt.grid(True)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/pca_cluster_visualization.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 23. CLUSTER DISTRIBUTION
# ============================================================

cluster_distribution = (
    customer_df[
        "Cluster"
    ]
    .value_counts()
    .sort_index()
    .reset_index()
)

cluster_distribution.columns = [
    "Cluster",
    "Customer_Count"
]

cluster_distribution[
    "Percentage"
] = (
    cluster_distribution[
        "Customer_Count"
    ]
    / len(customer_df)
    * 100
)

print(
    "\nCluster Distribution:"
)

display(
    cluster_distribution
)

cluster_distribution.to_csv(
    f"{OUTPUT_DIR}/cluster_distribution.csv",
    index=False
)


# ============================================================
# 24. CLUSTER PROFILE
# ============================================================

cluster_profile = (
    customer_df
    .groupby("Cluster")[features]
    .mean()
)

print(
    "\nCustomer Cluster Profile:"
)

display(
    cluster_profile
)

cluster_profile.to_csv(
    f"{OUTPUT_DIR}/cluster_profile.csv"
)


# ============================================================
# 25. t-SNE PERPLEXITY EXPERIMENT
# ============================================================

perplexity_values = [
    5,
    10,
    20,
    30,
    40,
    50
]

perplexity_results = []

for perplexity in perplexity_values:

    tsne_model = TSNE(
        n_components=2,
        perplexity=perplexity,
        learning_rate="auto",
        init="pca",
        max_iter=750,
        random_state=RANDOM_STATE
    )

    embedding = tsne_model.fit_transform(
        X_scaled
    )

    perplexity_results.append({

        "Perplexity": perplexity,

        "KL_Divergence":
            tsne_model.kl_divergence_
    })


perplexity_df = pd.DataFrame(
    perplexity_results
)

print(
    "\nt-SNE Perplexity Analysis:"
)

display(
    perplexity_df
)

perplexity_df.to_csv(
    f"{OUTPUT_DIR}/tsne_perplexity_analysis.csv",
    index=False
)


# ============================================================
# 26. SAVE MODELS
# ============================================================

joblib.dump(
    scaler,
    f"{OUTPUT_DIR}/standard_scaler.pkl"
)

joblib.dump(
    pca,
    f"{OUTPUT_DIR}/pca_model.pkl"
)

joblib.dump(
    final_kmeans,
    f"{OUTPUT_DIR}/kmeans_model.pkl"
)

print(
    "\nModels saved successfully."
)


# ============================================================
# 27. NEW CUSTOMER
# ============================================================

new_customer = pd.DataFrame({

    "Age": [30],

    "Annual_Income": [70000],

    "Purchase_Frequency": [15],

    "Average_Order_Value": [100],

    "Website_Visits": [65],

    "Discount_Usage": [30],

    "Customer_Lifetime_Value": [4500],

    "App_Usage_Hours": [18],

    "Support_Requests": [4],

    "Product_Reviews": [10]
})


# ============================================================
# 28. NEW CUSTOMER TRANSFORMATION
# ============================================================

new_scaled = scaler.transform(
    new_customer[features]
)

new_pca = pca.transform(
    new_scaled
)

predicted_cluster = (
    final_kmeans.predict(
        new_scaled
    )[0]
)

print(
    "\nNew Customer Predicted Cluster:",
    predicted_cluster
)

print(
    "PCA Coordinates:"
)

print(
    new_pca
)


# ============================================================
# 29. SAVE NEW CUSTOMER RESULT
# ============================================================

new_customer[
    "Predicted_Cluster"
] = predicted_cluster

new_customer.to_csv(
    f"{OUTPUT_DIR}/new_customer_prediction.csv",
    index=False
)


# ============================================================
# 30. SAVE FINAL DATASET
# ============================================================

final_dataset = customer_df[
    [
        "Customer_ID"
    ]
    + features
    + [
        "PCA_1",
        "PCA_2",
        "TSNE_1",
        "TSNE_2",
        "Cluster"
    ]
]

final_dataset.to_csv(
    f"{OUTPUT_DIR}/final_customer_dataset.csv",
    index=False
)


# ============================================================
# 31. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 65)

print(
    "DAY 28 - t-SNE CUSTOMER SEGMENTATION"
)

print("=" * 65)

print(
    "\nOriginal features:",
    len(features)
)

print(
    "PCA dimensions:",
    X_pca.shape[1]
)

print(
    "t-SNE dimensions:",
    X_tsne.shape[1]
)

print(
    "Selected K:",
    best_k
)

print(
    "Silhouette Score:",
    round(
        final_silhouette,
        4
    )
)

print(
    "\nGenerated files:"
)

for file in sorted(
    os.listdir(OUTPUT_DIR)
):
    print(
        "-",
        file
    )
```
