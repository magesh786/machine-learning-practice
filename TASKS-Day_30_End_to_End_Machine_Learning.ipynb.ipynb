```python
# ============================================================
# DAY 30 - END-TO-END MACHINE LEARNING PROJECT
# Customer Churn Prediction
# Random Forest vs XGBoost
# ============================================================


# ============================================================
# 1. INSTALL REQUIRED LIBRARIES
# ============================================================

!pip -q install xgboost joblib


# ============================================================
# 2. IMPORT LIBRARIES
# ============================================================

import os
import warnings
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_val_score,
    RandomizedSearchCV
)

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    RocCurveDisplay
)

from xgboost import XGBClassifier

warnings.filterwarnings("ignore")

RANDOM_STATE = 42

OUTPUT_DIR = "day30_outputs"

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

np.random.seed(
    RANDOM_STATE
)

print("Libraries imported successfully.")


# ============================================================
# 3. CREATE DATASET
# ============================================================

n_customers = 5000

data = pd.DataFrame({

    "Age": np.random.randint(
        18,
        70,
        n_customers
    ),

    "Tenure_Months": np.random.randint(
        1,
        72,
        n_customers
    ),

    "Monthly_Charges": np.random.uniform(
        20,
        150,
        n_customers
    ).round(2),

    "Support_Calls": np.random.poisson(
        3,
        n_customers
    ),

    "Login_Frequency": np.random.randint(
        1,
        31,
        n_customers
    ),

    "Usage_Hours": np.random.uniform(
        1,
        80,
        n_customers
    ).round(2),

    "Discount_Usage": np.random.uniform(
        0,
        100,
        n_customers
    ).round(2),

    "Payment_Delays": np.random.poisson(
        2,
        n_customers
    ),

    "Satisfaction_Score": np.random.randint(
        1,
        11,
        n_customers
    ),

    "Total_Spend": np.random.uniform(
        100,
        15000,
        n_customers
    ).round(2)
})


# ============================================================
# 4. GENERATE CHURN TARGET
# ============================================================

churn_signal = (

    0.040 * data["Support_Calls"]

    + 0.055 * data["Payment_Delays"]

    + 0.010 * data["Monthly_Charges"]

    - 0.020 * data["Tenure_Months"]

    - 0.12 * data["Satisfaction_Score"]

    - 0.018 * data["Login_Frequency"]

    - 0.005 * data["Usage_Hours"]

)

churn_signal = (
    churn_signal -
    churn_signal.mean()
)

probability = (
    1 /
    (
        1 +
        np.exp(
            -churn_signal
        )
    )
)

data["Churn"] = (
    np.random.random(
        n_customers
    )
    < probability
).astype(int)


# ============================================================
# 5. CUSTOMER ID
# ============================================================

data.insert(
    0,
    "Customer_ID",
    range(
        1,
        n_customers + 1
    )
)


# ============================================================
# 6. DISPLAY DATASET
# ============================================================

print(
    "Dataset shape:",
    data.shape
)

display(
    data.head()
)


# ============================================================
# 7. DATA INFORMATION
# ============================================================

print("\nDataset Information:")

data.info()


# ============================================================
# 8. MISSING VALUES
# ============================================================

print(
    "\nMissing Values:"
)

print(
    data.isnull().sum()
)


# ============================================================
# 9. DUPLICATES
# ============================================================

print(
    "\nDuplicate Rows:",
    data.duplicated().sum()
)


# ============================================================
# 10. CLASS DISTRIBUTION
# ============================================================

print(
    "\nChurn Distribution:"
)

print(
    data["Churn"].value_counts()
)


# ============================================================
# 11. CLASS DISTRIBUTION PLOT
# ============================================================

data["Churn"].value_counts().plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title(
    "Customer Churn Distribution"
)

plt.xlabel(
    "Churn"
)

plt.ylabel(
    "Number of Customers"
)

plt.xticks(
    rotation=0
)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/churn_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 12. FEATURES AND TARGET
# ============================================================

features = [

    "Age",

    "Tenure_Months",

    "Monthly_Charges",

    "Support_Calls",

    "Login_Frequency",

    "Usage_Hours",

    "Discount_Usage",

    "Payment_Delays",

    "Satisfaction_Score",

    "Total_Spend"

]

X = data[
    features
]

y = data[
    "Churn"
]


# ============================================================
# 13. TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=RANDOM_STATE,

    stratify=y
)


print(
    "\nTraining samples:",
    len(X_train)
)

print(
    "Testing samples:",
    len(X_test)
)


# ============================================================
# 14. RANDOM FOREST MODEL
# ============================================================

rf_model = RandomForestClassifier(

    n_estimators=300,

    max_depth=None,

    min_samples_split=2,

    min_samples_leaf=1,

    max_features="sqrt",

    class_weight="balanced",

    random_state=RANDOM_STATE,

    n_jobs=-1
)


rf_model.fit(
    X_train,
    y_train
)

print(
    "\nRandom Forest trained."
)


# ============================================================
# 15. RANDOM FOREST PREDICTION
# ============================================================

rf_pred = rf_model.predict(
    X_test
)

rf_probability = (
    rf_model.predict_proba(
        X_test
    )[:, 1]
)


# ============================================================
# 16. RANDOM FOREST METRICS
# ============================================================

rf_accuracy = accuracy_score(
    y_test,
    rf_pred
)

rf_precision = precision_score(
    y_test,
    rf_pred
)

rf_recall = recall_score(
    y_test,
    rf_pred
)

rf_f1 = f1_score(
    y_test,
    rf_pred
)

rf_auc = roc_auc_score(
    y_test,
    rf_probability
)


rf_metrics = {

    "Accuracy": rf_accuracy,

    "Precision": rf_precision,

    "Recall": rf_recall,

    "F1 Score": rf_f1,

    "ROC-AUC": rf_auc

}


print(
    "\nRandom Forest Metrics:"
)

for metric, score in rf_metrics.items():

    print(
        f"{metric}: {score:.4f}"
    )


# ============================================================
# 17. RANDOM FOREST REPORT
# ============================================================

print(
    "\nRandom Forest Classification Report:"
)

print(
    classification_report(
        y_test,
        rf_pred
    )
)


# ============================================================
# 18. XGBOOST MODEL
# ============================================================

xgb_model = XGBClassifier(

    n_estimators=300,

    max_depth=5,

    learning_rate=0.05,

    subsample=0.8,

    colsample_bytree=0.8,

    objective="binary:logistic",

    eval_metric="logloss",

    random_state=RANDOM_STATE,

    n_jobs=-1
)


xgb_model.fit(
    X_train,
    y_train
)

print(
    "\nXGBoost trained."
)


# ============================================================
# 19. XGBOOST PREDICTION
# ============================================================

xgb_pred = xgb_model.predict(
    X_test
)

xgb_probability = (
    xgb_model.predict_proba(
        X_test
    )[:, 1]
)


# ============================================================
# 20. XGBOOST METRICS
# ============================================================

xgb_accuracy = accuracy_score(
    y_test,
    xgb_pred
)

xgb_precision = precision_score(
    y_test,
    xgb_pred
)

xgb_recall = recall_score(
    y_test,
    xgb_pred
)

xgb_f1 = f1_score(
    y_test,
    xgb_pred
)

xgb_auc = roc_auc_score(
    y_test,
    xgb_probability
)


xgb_metrics = {

    "Accuracy": xgb_accuracy,

    "Precision": xgb_precision,

    "Recall": xgb_recall,

    "F1 Score": xgb_f1,

    "ROC-AUC": xgb_auc

}


print(
    "\nXGBoost Metrics:"
)

for metric, score in xgb_metrics.items():

    print(
        f"{metric}: {score:.4f}"
    )


# ============================================================
# 21. XGBOOST REPORT
# ============================================================

print(
    "\nXGBoost Classification Report:"
)

print(
    classification_report(
        y_test,
        xgb_pred
    )
)


# ============================================================
# 22. MODEL COMPARISON
# ============================================================

comparison = pd.DataFrame({

    "Metric": [

        "Accuracy",

        "Precision",

        "Recall",

        "F1 Score",

        "ROC-AUC"

    ],

    "Random Forest": [

        rf_accuracy,

        rf_precision,

        rf_recall,

        rf_f1,

        rf_auc

    ],

    "XGBoost": [

        xgb_accuracy,

        xgb_precision,

        xgb_recall,

        xgb_f1,

        xgb_auc

    ]

})


print(
    "\nModel Comparison:"
)

display(
    comparison
)


# ============================================================
# 23. SAVE COMPARISON
# ============================================================

comparison.to_csv(

    f"{OUTPUT_DIR}/model_comparison.csv",

    index=False
)


# ============================================================
# 24. CROSS-VALIDATION
# ============================================================

cv = StratifiedKFold(

    n_splits=5,

    shuffle=True,

    random_state=RANDOM_STATE
)


rf_cv_scores = cross_val_score(

    rf_model,

    X,

    y,

    cv=cv,

    scoring="roc_auc",

    n_jobs=-1
)


xgb_cv_scores = cross_val_score(

    xgb_model,

    X,

    y,

    cv=cv,

    scoring="roc_auc",

    n_jobs=-1
)


print(
    "\nRandom Forest CV ROC-AUC:"
)

print(
    rf_cv_scores
)

print(
    "Mean:",
    rf_cv_scores.mean()
)


print(
    "\nXGBoost CV ROC-AUC:"
)

print(
    xgb_cv_scores
)

print(
    "Mean:",
    xgb_cv_scores.mean()
)


# ============================================================
# 25. CROSS VALIDATION TABLE
# ============================================================

cv_results = pd.DataFrame({

    "Model": [

        "Random Forest",

        "XGBoost"

    ],

    "Mean_CV_ROC_AUC": [

        rf_cv_scores.mean(),

        xgb_cv_scores.mean()

    ],

    "Std_CV_ROC_AUC": [

        rf_cv_scores.std(),

        xgb_cv_scores.std()

    ]

})


display(
    cv_results
)


cv_results.to_csv(

    f"{OUTPUT_DIR}/cross_validation_results.csv",

    index=False
)


# ============================================================
# 26. ROC CURVE COMPARISON
# ============================================================

plt.figure(
    figsize=(10, 7)
)

RocCurveDisplay.from_predictions(

    y_test,

    rf_probability,

    name="Random Forest"

)

RocCurveDisplay.from_predictions(

    y_test,

    xgb_probability,

    name="XGBoost"

)

plt.title(
    "Random Forest vs XGBoost ROC Curve"
)

plt.grid(True)

plt.tight_layout()

plt.savefig(

    f"{OUTPUT_DIR}/roc_comparison.png",

    dpi=300,

    bbox_inches="tight"

)

plt.show()


# ============================================================
# 27. RANDOM FOREST CONFUSION MATRIX
# ============================================================

rf_cm = confusion_matrix(

    y_test,

    rf_pred

)

ConfusionMatrixDisplay(

    confusion_matrix=rf_cm

).plot()

plt.title(
    "Random Forest Confusion Matrix"
)

plt.tight_layout()

plt.savefig(

    f"{OUTPUT_DIR}/rf_confusion_matrix.png",

    dpi=300,

    bbox_inches="tight"

)

plt.show()


# ============================================================
# 28. XGBOOST CONFUSION MATRIX
# ============================================================

xgb_cm = confusion_matrix(

    y_test,

    xgb_pred

)

ConfusionMatrixDisplay(

    confusion_matrix=xgb_cm

).plot()

plt.title(
    "XGBoost Confusion Matrix"
)

plt.tight_layout()

plt.savefig(

    f"{OUTPUT_DIR}/xgb_confusion_matrix.png",

    dpi=300,

    bbox_inches="tight"

)

plt.show()


# ============================================================
# 29. RANDOM FOREST FEATURE IMPORTANCE
# ============================================================

rf_importance = pd.DataFrame({

    "Feature": features,

    "Importance":
        rf_model.feature_importances_

})


rf_importance = (

    rf_importance

    .sort_values(

        "Importance",

        ascending=False

    )

)


display(
    rf_importance
)


# ============================================================
# 30. XGBOOST FEATURE IMPORTANCE
# ============================================================

xgb_importance = pd.DataFrame({

    "Feature": features,

    "Importance":
        xgb_model.feature_importances_

})


xgb_importance = (

    xgb_importance

    .sort_values(

        "Importance",

        ascending=False

    )

)


display(
    xgb_importance
)


# ============================================================
# 31. FEATURE IMPORTANCE COMPARISON
# ============================================================

feature_comparison = pd.DataFrame({

    "Feature": features,

    "Random_Forest":

        rf_model.feature_importances_,

    "XGBoost":

        xgb_model.feature_importances_

})


display(
    feature_comparison
)


feature_comparison.to_csv(

    f"{OUTPUT_DIR}/feature_importance_comparison.csv",

    index=False

)


# ============================================================
# 32. XGBOOST FEATURE IMPORTANCE PLOT
# ============================================================

plt.figure(
    figsize=(10, 7)
)

plt.barh(

    xgb_importance[
        "Feature"
    ],

    xgb_importance[
        "Importance"
    ]

)

plt.xlabel(
    "Importance"
)

plt.ylabel(
    "Feature"
)

plt.title(
    "XGBoost Feature Importance"
)

plt.gca().invert_yaxis()

plt.tight_layout()

plt.savefig(

    f"{OUTPUT_DIR}/xgb_feature_importance.png",

    dpi=300,

    bbox_inches="tight"

)

plt.show()


# ============================================================
# 33. SELECT MODEL FOR DEPLOYMENT
# ============================================================

if xgb_auc >= rf_auc:

    final_model = xgb_model

    final_model_name = "XGBoost"

else:

    final_model = rf_model

    final_model_name = "Random Forest"


print(
    "\nSelected deployment model:",
    final_model_name
)


# ============================================================
# 34. SAVE FINAL MODEL
# ============================================================

joblib.dump(

    final_model,

    f"{OUTPUT_DIR}/final_churn_model.pkl"

)


# ============================================================
# 35. SAVE FEATURE LIST
# ============================================================

joblib.dump(

    features,

    f"{OUTPUT_DIR}/feature_list.pkl"

)


# ============================================================
# 36. CREATE NEW CUSTOMER
# ============================================================

new_customer = pd.DataFrame({

    "Age": [28],

    "Tenure_Months": [7],

    "Monthly_Charges": [105],

    "Support_Calls": [7],

    "Login_Frequency": [6],

    "Usage_Hours": [10],

    "Discount_Usage": [15],

    "Payment_Delays": [5],

    "Satisfaction_Score": [3],

    "Total_Spend": [950]

})


# ============================================================
# 37. NEW CUSTOMER PREDICTION
# ============================================================

new_prediction = (

    final_model.predict(

        new_customer

    )[0]

)


new_probability = (

    final_model.predict_proba(

        new_customer

    )[0][1]

)


print(
    "\nNew Customer Prediction"
)

print(
    "Prediction:",
    new_prediction
)

print(
    "Churn Probability:",
    round(
        new_probability,
        4
    )
)


# ============================================================
# 38. ADD RESULT
# ============================================================

new_customer[
    "Predicted_Churn"
] = new_prediction

new_customer[
    "Churn_Probability"
] = new_probability


display(
    new_customer
)


# ============================================================
# 39. SAVE PREDICTION
# ============================================================

new_customer.to_csv(

    f"{OUTPUT_DIR}/new_customer_prediction.csv",

    index=False

)


# ============================================================
# 40. SAVE DATASET
# ============================================================

data.to_csv(

    f"{OUTPUT_DIR}/customer_churn_dataset.csv",

    index=False

)


# ============================================================
# 41. FINAL PROJECT SUMMARY
# ============================================================

print("\n" + "=" * 70)

print(
    "DAY 30 - END-TO-END MACHINE LEARNING"
)

print("=" * 70)

print(
    "\nRandom Forest Test ROC-AUC:",
    round(
        rf_auc,
        4
    )
)

print(
    "XGBoost Test ROC-AUC:",
    round(
        xgb_auc,
        4
    )
)

print(
    "\nSelected Model:",
    final_model_name
)

print(
    "New Customer Churn Probability:",
    round(
        new_probability,
        4
    )
)

print(
    "\nProject completed successfully!"
)

print(
    "\nGenerated files:"
)

for file in sorted(
    os.listdir(
        OUTPUT_DIR
    )
):

    print(
        "-",
        file
    )
```
