```python
# ============================================================
# DAY 27 - PCA DIMENSIONALITY REDUCTION + CUSTOMER SEGMENTATION
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import warnings
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import (
    silhouette_score,
    calinski_harabasz_score,
    davies_bouldin_score
)

warnings.filterwarnings("ignore")

RANDOM_STATE = 42

OUTPUT_DIR = "day27_outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

np.random.seed(RANDOM_STATE)

print("Libraries imported successfully.")


# ============================================================
# 2. CREATE CUSTOMER DATASET
# ============================================================

n_customers = 1500

customer_df = pd.DataFrame({

    "Age": np.random.normal(
        35, 10, n_customers
    ),

    "Annual_Income": np.random.normal(
        65000, 22000, n_customers
    ),

    "Purchase_Frequency": np.random.normal(
        12, 5, n_customers
    ),

    "Average_Order_Value": np.random.normal(
        90, 30, n_customers
    ),

    "Website_Visits": np.random.normal(
        55, 20, n_customers
    ),

    "Discount_Usage": np.random.normal(
        45, 20, n_customers
    ),

    "Customer_Lifetime_Value": np.random.normal(
        3500, 1200, n_customers
    ),

    "App_Usage_Hours": np.random.normal(
        15, 7, n_customers
    ),

    "Support_Requests": np.random.normal(
        5, 3, n_customers
    ),

    "Product_Reviews": np.random.normal(
        8, 4, n_customers
    )
})

# ------------------------------------------------------------
# Clip unrealistic values
# ------------------------------------------------------------

customer_df["Age"] = customer_df[
    "Age"
].clip(18, 75)

customer_df["Annual_Income"] = customer_df[
    "Annual_Income"
].clip(15000, 200000)

customer_df["Purchase_Frequency"] = customer_df[
    "Purchase_Frequency"
].clip(1, 50)

customer_df["Average_Order_Value"] = customer_df[
    "Average_Order_Value"
].clip(5, 500)

customer_df["Website_Visits"] = customer_df[
    "Website_Visits"
].clip(1, 200)

customer_df["Discount_Usage"] = customer_df[
    "Discount_Usage"
].clip(0, 100)

customer_df["Customer_Lifetime_Value"] = customer_df[
    "Customer_Lifetime_Value"
].clip(100, 15000)

customer_df["App_Usage_Hours"] = customer_df[
    "App_Usage_Hours"
].clip(0, 60)

customer_df["Support_Requests"] = customer_df[
    "Support_Requests"
].clip(0, 30)

customer_df["Product_Reviews"] = customer_df[
    "Product_Reviews"
].clip(0, 30)

customer_df = customer_df.round(2)

customer_df.insert(
    0,
    "Customer_ID",
    range(1, len(customer_df) + 1)
)

print("Dataset shape:", customer_df.shape)

display(customer_df.head())


# ============================================================
# 3. DATA EXPLORATION
# ============================================================

print("\nDataset Information:")
customer_df.info()

print("\nMissing Values:")
print(customer_df.isnull().sum())

print("\nStatistical Summary:")
display(customer_df.describe())


# ============================================================
# 4. SAVE RAW DATA
# ============================================================

customer_df.to_csv(
    f"{OUTPUT_DIR}/customer_dataset.csv",
    index=False
)


# ============================================================
# 5. SELECT FEATURES
# ============================================================

features = [
    "Age",
    "Annual_Income",
    "Purchase_Frequency",
    "Average_Order_Value",
    "Website_Visits",
    "Discount_Usage",
    "Customer_Lifetime_Value",
    "App_Usage_Hours",
    "Support_Requests",
    "Product_Reviews"
]

X = customer_df[features].copy()

print("\nNumber of original features:")
print(X.shape[1])


# ============================================================
# 6. CHECK CORRELATION
# ============================================================

correlation_matrix = X.corr()

print("\nCorrelation Matrix:")
display(correlation_matrix)

correlation_matrix.to_csv(
    f"{OUTPUT_DIR}/correlation_matrix.csv"
)


# ============================================================
# 7. STANDARDIZE FEATURES
# ============================================================

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print(
    "\nStandardization completed."
)


# ============================================================
# 8. PCA WITH ALL COMPONENTS
# ============================================================

pca_full = PCA(
    random_state=RANDOM_STATE
)

X_pca_full = pca_full.fit_transform(
    X_scaled
)

explained_variance = (
    pca_full.explained_variance_ratio_
)

cumulative_variance = np.cumsum(
    explained_variance
)

pca_variance_df = pd.DataFrame({

    "Component": np.arange(
        1,
        len(explained_variance) + 1
    ),

    "Explained_Variance":
        explained_variance,

    "Cumulative_Variance":
        cumulative_variance
})

print("\nPCA Explained Variance:")
display(pca_variance_df)

pca_variance_df.to_csv(
    f"{OUTPUT_DIR}/pca_explained_variance.csv",
    index=False
)


# ============================================================
# 9. SCREE PLOT
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    pca_variance_df["Component"],
    pca_variance_df[
        "Explained_Variance"
    ],
    marker="o"
)

plt.xlabel("Principal Component")
plt.ylabel("Explained Variance Ratio")

plt.title(
    "PCA Scree Plot"
)

plt.xticks(
    pca_variance_df["Component"]
)

plt.grid(True)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/pca_scree_plot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 10. CUMULATIVE EXPLAINED VARIANCE
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    pca_variance_df["Component"],
    pca_variance_df[
        "Cumulative_Variance"
    ],
    marker="o"
)

plt.axhline(
    y=0.90,
    linestyle="--",
    label="90% Variance"
)

plt.axhline(
    y=0.95,
    linestyle="--",
    label="95% Variance"
)

plt.xlabel("Number of Components")
plt.ylabel("Cumulative Explained Variance")

plt.title(
    "PCA Cumulative Explained Variance"
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/pca_cumulative_variance.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 11. DETERMINE COMPONENTS FOR 90% AND 95%
# ============================================================

components_90 = np.argmax(
    cumulative_variance >= 0.90
) + 1

components_95 = np.argmax(
    cumulative_variance >= 0.95
) + 1

print(
    "Components required for 90% variance:",
    components_90
)

print(
    "Components required for 95% variance:",
    components_95
)


# ============================================================
# 12. PCA WITH SELECTED COMPONENTS
# ============================================================

# Use the number of components needed
# to preserve at least 90% variance.

pca = PCA(
    n_components=components_90,
    random_state=RANDOM_STATE
)

X_reduced = pca.fit_transform(
    X_scaled
)

print(
    "\nOriginal dimensions:",
    X_scaled.shape[1]
)

print(
    "Reduced dimensions:",
    X_reduced.shape[1]
)

print(
    "Variance retained:",
    round(
        pca.explained_variance_ratio_.sum(),
        4
    )
)


# ============================================================
# 13. CREATE PCA DATAFRAME
# ============================================================

pca_columns = [
    f"PC{i+1}"
    for i in range(
        X_reduced.shape[1]
    )
]

pca_df = pd.DataFrame(
    X_reduced,
    columns=pca_columns
)

pca_df.insert(
    0,
    "Customer_ID",
    customer_df["Customer_ID"]
)

display(pca_df.head())

pca_df.to_csv(
    f"{OUTPUT_DIR}/pca_reduced_dataset.csv",
    index=False
)


# ============================================================
# 14. PCA COMPONENT LOADINGS
# ============================================================

loadings = pd.DataFrame(
    pca.components_.T,
    index=features,
    columns=pca_columns
)

print("\nPCA Component Loadings:")
display(loadings)

loadings.to_csv(
    f"{OUTPUT_DIR}/pca_component_loadings.csv"
)


# ============================================================
# 15. FIND TOP FEATURES FOR EACH COMPONENT
# ============================================================

for component in loadings.columns:

    print(
        f"\nTop features for {component}:"
    )

    top_features = (
        loadings[component]
        .abs()
        .sort_values(
            ascending=False
        )
        .head(5)
    )

    for feature in top_features.index:

        value = loadings.loc[
            feature,
            component
        ]

        print(
            f"{feature}: {value:.4f}"
        )


# ============================================================
# 16. 2D PCA FOR VISUALIZATION
# ============================================================

pca_2d = PCA(
    n_components=2,
    random_state=RANDOM_STATE
)

X_pca_2d = pca_2d.fit_transform(
    X_scaled
)

customer_df["PC1"] = X_pca_2d[:, 0]
customer_df["PC2"] = X_pca_2d[:, 1]


# ============================================================
# 17. 2D PCA VISUALIZATION
# ============================================================

plt.figure(figsize=(12, 8))

plt.scatter(
    customer_df["PC1"],
    customer_df["PC2"],
    alpha=0.6
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.title(
    "Customer Data in PCA 2D Space"
)

plt.grid(True)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/pca_2d_visualization.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 18. K-MEANS ON PCA DATA
# ============================================================

k_values = range(2, 9)

kmeans_results = []

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=RANDOM_STATE,
        n_init=20
    )

    labels = model.fit_predict(
        X_reduced
    )

    silhouette = silhouette_score(
        X_reduced,
        labels
    )

    inertia = model.inertia_

    calinski = calinski_harabasz_score(
        X_reduced,
        labels
    )

    davies = davies_bouldin_score(
        X_reduced,
        labels
    )

    kmeans_results.append({

        "K": k,

        "Inertia": inertia,

        "Silhouette_Score":
            silhouette,

        "Calinski_Harabasz":
            calinski,

        "Davies_Bouldin":
            davies
    })


kmeans_results_df = pd.DataFrame(
    kmeans_results
)

print(
    "\nK-Means Results on PCA Data:"
)

display(
    kmeans_results_df
)

kmeans_results_df.to_csv(
    f"{OUTPUT_DIR}/kmeans_pca_analysis.csv",
    index=False
)


# ============================================================
# 19. ELBOW CURVE
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    kmeans_results_df["K"],
    kmeans_results_df["Inertia"],
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")

plt.title(
    "Elbow Method after PCA"
)

plt.grid(True)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/pca_kmeans_elbow.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 20. SILHOUETTE ANALYSIS
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    kmeans_results_df["K"],
    kmeans_results_df[
        "Silhouette_Score"
    ],
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")

plt.title(
    "Silhouette Analysis after PCA"
)

plt.grid(True)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/pca_kmeans_silhouette.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 21. SELECT BEST K
# ============================================================

best_k = int(
    kmeans_results_df.loc[
        kmeans_results_df[
            "Silhouette_Score"
        ].idxmax(),
        "K"
    ]
)

print(
    "\nSelected K based on silhouette score:",
    best_k
)


# ============================================================
# 22. TRAIN FINAL K-MEANS
# ============================================================

final_kmeans = KMeans(
    n_clusters=best_k,
    random_state=RANDOM_STATE,
    n_init=30
)

final_labels = final_kmeans.fit_predict(
    X_reduced
)

customer_df[
    "PCA_Cluster"
] = final_labels


# ============================================================
# 23. FINAL CLUSTER METRICS
# ============================================================

final_silhouette = silhouette_score(
    X_reduced,
    final_labels
)

final_calinski = calinski_harabasz_score(
    X_reduced,
    final_labels
)

final_davies = davies_bouldin_score(
    X_reduced,
    final_labels
)

final_metrics = pd.DataFrame({

    "Metric": [

        "Original Features",

        "PCA Components",

        "Variance Retained",

        "Number of Clusters",

        "Silhouette Score",

        "Calinski-Harabasz Score",

        "Davies-Bouldin Score",

        "Inertia"
    ],

    "Value": [

        len(features),

        X_reduced.shape[1],

        pca.explained_variance_ratio_.sum(),

        best_k,

        final_silhouette,

        final_calinski,

        final_davies,

        final_kmeans.inertia_
    ]
})

print(
    "\nFinal Model Metrics:"
)

display(final_metrics)

final_metrics.to_csv(
    f"{OUTPUT_DIR}/final_metrics.csv",
    index=False
)


# ============================================================
# 24. PCA + K-MEANS VISUALIZATION
# ============================================================

plt.figure(figsize=(12, 8))

scatter = plt.scatter(
    customer_df["PC1"],
    customer_df["PC2"],
    c=customer_df["PCA_Cluster"],
    alpha=0.7
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.title(
    "PCA + K-Means Customer Segmentation"
)

plt.colorbar(
    scatter,
    label="Cluster"
)

plt.grid(True)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/pca_kmeans_clusters.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 25. CLUSTER DISTRIBUTION
# ============================================================

cluster_distribution = (
    customer_df[
        "PCA_Cluster"
    ]
    .value_counts()
    .sort_index()
    .reset_index()
)

cluster_distribution.columns = [
    "Cluster",
    "Customer_Count"
]

cluster_distribution[
    "Percentage"
] = (
    cluster_distribution[
        "Customer_Count"
    ]
    / len(customer_df)
    * 100
)

print(
    "\nCluster Distribution:"
)

display(
    cluster_distribution
)

cluster_distribution.to_csv(
    f"{OUTPUT_DIR}/cluster_distribution.csv",
    index=False
)


# ============================================================
# 26. CLUSTER PROFILE USING ORIGINAL FEATURES
# ============================================================

cluster_profile = (
    customer_df
    .groupby("PCA_Cluster")[features]
    .mean()
)

print(
    "\nCluster Profile:"
)

display(
    cluster_profile
)

cluster_profile.to_csv(
    f"{OUTPUT_DIR}/cluster_profile.csv"
)


# ============================================================
# 27. PCA CLUSTER CENTERS
# ============================================================

cluster_centers = pd.DataFrame(
    final_kmeans.cluster_centers_,
    columns=pca_columns
)

cluster_centers.insert(
    0,
    "Cluster",
    range(best_k)
)

print(
    "\nCluster Centers in PCA Space:"
)

display(
    cluster_centers
)

cluster_centers.to_csv(
    f"{OUTPUT_DIR}/pca_cluster_centers.csv",
    index=False
)


# ============================================================
# 28. ORIGINAL SPACE APPROXIMATION OF CENTERS
# ============================================================

centers_original_scaled = (
    final_kmeans.cluster_centers_
    @ pca.components_
)

centers_original = (
    scaler.inverse_transform(
        centers_original_scaled
    )
)

centers_original_df = pd.DataFrame(
    centers_original,
    columns=features
)

centers_original_df.insert(
    0,
    "Cluster",
    range(best_k)
)

print(
    "\nApproximate Cluster Centers "
    "in Original Feature Space:"
)

display(
    centers_original_df
)

centers_original_df.to_csv(
    f"{OUTPUT_DIR}/cluster_centers_original_space.csv",
    index=False
)


# ============================================================
# 29. COMPARE ORIGINAL SPACE VS PCA SPACE
# ============================================================

original_kmeans = KMeans(
    n_clusters=best_k,
    random_state=RANDOM_STATE,
    n_init=30
)

original_labels = (
    original_kmeans.fit_predict(
        X_scaled
    )
)

original_silhouette = (
    silhouette_score(
        X_scaled,
        original_labels
    )
)

comparison_df = pd.DataFrame({

    "Approach": [
        "K-Means on Original Features",
        "PCA + K-Means"
    ],

    "Number_of_Features": [
        X_scaled.shape[1],
        X_reduced.shape[1]
    ],

    "Silhouette_Score": [
        original_silhouette,
        final_silhouette
    ],

    "Inertia": [
        original_kmeans.inertia_,
        final_kmeans.inertia_
    ]
})

print(
    "\nOriginal vs PCA Comparison:"
)

display(
    comparison_df
)

comparison_df.to_csv(
    f"{OUTPUT_DIR}/original_vs_pca_comparison.csv",
    index=False
)


# ============================================================
# 30. SAVE MODELS
# ============================================================

joblib.dump(
    scaler,
    f"{OUTPUT_DIR}/standard_scaler.pkl"
)

joblib.dump(
    pca,
    f"{OUTPUT_DIR}/pca_model.pkl"
)

joblib.dump(
    final_kmeans,
    f"{OUTPUT_DIR}/pca_kmeans_model.pkl"
)

print(
    "\nModels saved successfully."
)


# ============================================================
# 31. NEW CUSTOMER TRANSFORMATION
# ============================================================

new_customer = pd.DataFrame({

    "Age": [30],

    "Annual_Income": [70000],

    "Purchase_Frequency": [15],

    "Average_Order_Value": [100],

    "Website_Visits": [65],

    "Discount_Usage": [30],

    "Customer_Lifetime_Value": [4500],

    "App_Usage_Hours": [18],

    "Support_Requests": [4],

    "Product_Reviews": [10]
})

# Standardize
new_customer_scaled = scaler.transform(
    new_customer[features]
)

# PCA transformation
new_customer_pca = pca.transform(
    new_customer_scaled
)

# Predict cluster
new_customer_cluster = (
    final_kmeans.predict(
        new_customer_pca
    )[0]
)

print(
    "\nNEW CUSTOMER PREDICTION"
)

print(
    "Predicted Cluster:",
    new_customer_cluster
)

print(
    "PCA Coordinates:"
)

print(
    new_customer_pca
)


# ============================================================
# 32. SAVE NEW CUSTOMER RESULT
# ============================================================

new_customer["Predicted_Cluster"] = (
    new_customer_cluster
)

new_customer.to_csv(
    f"{OUTPUT_DIR}/new_customer_prediction.csv",
    index=False
)


# ============================================================
# 33. SAVE FINAL DATASET
# ============================================================

final_customer_data = customer_df[
    [
        "Customer_ID"
    ] + features + [
        "PC1",
        "PC2",
        "PCA_Cluster"
    ]
]

final_customer_data.to_csv(
    f"{OUTPUT_DIR}/final_customer_dataset.csv",
    index=False
)


# ============================================================
# 34. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 65)

print(
    "DAY 27 - PCA + K-MEANS "
    "CUSTOMER ANALYTICS COMPLETED"
)

print("=" * 65)

print(
    "\nOriginal number of features:",
    len(features)
)

print(
    "Reduced number of features:",
    X_reduced.shape[1]
)

print(
    "Variance retained:",
    round(
        pca.explained_variance_ratio_.sum(),
        4
    )
)

print(
    "Selected K:",
    best_k
)

print(
    "Silhouette Score:",
    round(
        final_silhouette,
        4
    )
)

print(
    "\nGenerated Files:"
)

for file in sorted(
    os.listdir(OUTPUT_DIR)
):

    print(
        "-",
        file
    )
```
