```python
# ============================================================
# DAY 10 - GRADIENT BOOSTING CLASSIFICATION
# Breast Cancer Dataset
# ============================================================

# 1. IMPORT LIBRARIES
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

print("Libraries imported successfully!")


# ============================================================
# 2. LOAD DATASET
# ============================================================

cancer = load_breast_cancer()

X = pd.DataFrame(
    cancer.data,
    columns=cancer.feature_names
)

y = pd.Series(
    cancer.target,
    name="Target"
)

df = X.copy()
df["Target"] = y

print("\nDataset loaded successfully!")
print("Dataset Shape:", df.shape)

display(df.head())


# ============================================================
# 3. DATASET INFORMATION
# ============================================================

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
display(df.describe())


# ============================================================
# 4. TARGET DISTRIBUTION
# ============================================================

print("\nTarget Distribution:")
print(df["Target"].value_counts())

print("\nTarget Names:")
for i, name in enumerate(cancer.target_names):
    print(i, "=", name)


# ============================================================
# 5. TARGET VISUALIZATION
# ============================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Target"
)

plt.title("Breast Cancer Target Distribution")
plt.xlabel("Target Class")
plt.ylabel("Number of Samples")
plt.show()


# ============================================================
# 6. SELECT FEATURES AND TARGET
# ============================================================

X = df.drop("Target", axis=1)
y = df["Target"]

print("\nNumber of Features:", X.shape[1])
print("Number of Samples:", X.shape[0])


# ============================================================
# 7. TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])


# ============================================================
# 8. FEATURE SCALING
# ============================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("\nFeature scaling completed!")


# ============================================================
# 9. CREATE GRADIENT BOOSTING MODEL
# ============================================================

model = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)

print("\nGradient Boosting model created!")
print(model)


# ============================================================
# 10. TRAIN MODEL
# ============================================================

model.fit(
    X_train_scaled,
    y_train
)

print("\nModel training completed!")


# ============================================================
# 11. MAKE PREDICTIONS
# ============================================================

y_pred = model.predict(X_test_scaled)

print("\nPredictions completed!")


# ============================================================
# 12. CALCULATE ACCURACY
# ============================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

print("\nModel Accuracy:")
print(round(accuracy * 100, 2), "%")


# ============================================================
# 13. CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=cancer.target_names
    )
)


# ============================================================
# 14. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred
)

print("\nConfusion Matrix:")
print(cm)


# ============================================================
# 15. CONFUSION MATRIX VISUALIZATION
# ============================================================

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=cancer.target_names
)

disp.plot()

plt.title("Gradient Boosting Confusion Matrix")
plt.show()


# ============================================================
# 16. FEATURE IMPORTANCE
# ============================================================

feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print("\nTop Important Features:")
display(feature_importance.head(15))


# ============================================================
# 17. FEATURE IMPORTANCE VISUALIZATION
# ============================================================

top_features = feature_importance.head(15)

plt.figure(figsize=(10, 7))

sns.barplot(
    data=top_features,
    x="Importance",
    y="Feature"
)

plt.title(
    "Top 15 Feature Importances - Gradient Boosting"
)

plt.xlabel("Importance")
plt.ylabel("Feature")

plt.show()


# ============================================================
# 18. PREDICTION PROBABILITY
# ============================================================

probabilities = model.predict_proba(
    X_test_scaled
)

print("\nPrediction Probabilities:")
display(
    pd.DataFrame(
        probabilities,
        columns=[
            "Malignant_Probability",
            "Benign_Probability"
        ]
    ).head(10)
)


# ============================================================
# 19. CREATE PREDICTION DATAFRAME
# ============================================================

prediction_df = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred,
    "Malignant_Probability": probabilities[:, 0],
    "Benign_Probability": probabilities[:, 1]
})

print("\nPrediction Results:")
display(prediction_df.head(15))


# ============================================================
# 20. ACTUAL VS PREDICTED
# ============================================================

comparison = prediction_df.copy()

comparison["Actual_Label"] = comparison[
    "Actual"
].map({
    0: "malignant",
    1: "benign"
})

comparison["Predicted_Label"] = comparison[
    "Predicted"
].map({
    0: "malignant",
    1: "benign"
})

print("\nActual vs Predicted:")
display(comparison.head(20))


# ============================================================
# 21. SAVE PREDICTIONS
# ============================================================

prediction_df.to_csv(
    "gradient_boosting_breast_cancer_predictions.csv",
    index=False
)

print(
    "\nPrediction results saved as:"
    "\ngradient_boosting_breast_cancer_predictions.csv"
)


# ============================================================
# 22. SAVE FEATURE IMPORTANCE
# ============================================================

feature_importance.to_csv(
    "gradient_boosting_feature_importance.csv",
    index=False
)

print(
    "\nFeature importance saved as:"
    "\ngradient_boosting_feature_importance.csv"
)


# ============================================================
# 23. SAVE MODEL SUMMARY
# ============================================================

model_summary = pd.DataFrame({
    "Metric": [
        "Algorithm",
        "Learning Type",
        "Training Samples",
        "Testing Samples",
        "Number of Features",
        "Number of Estimators",
        "Learning Rate",
        "Max Depth",
        "Accuracy"
    ],
    "Value": [
        "Gradient Boosting Classifier",
        "Supervised Learning",
        X_train.shape[0],
        X_test.shape[0],
        X.shape[1],
        100,
        0.1,
        3,
        round(accuracy, 4)
    ]
})

print("\nModel Summary:")
display(model_summary)


# ============================================================
# 24. SAVE MODEL SUMMARY
# ============================================================

model_summary.to_csv(
    "gradient_boosting_model_summary.csv",
    index=False
)

print("\nModel summary saved!")


# ============================================================
# 25. FINAL OUTPUT
# ============================================================

print("\n========================================")
print("DAY 10 PROJECT COMPLETED!")
print("========================================")
print("Project: Breast Cancer Classification")
print("Algorithm: Gradient Boosting")
print("Learning Type: Supervised Learning")
print("Number of Features:", X.shape[1])
print("Training Samples:", X_train.shape[0])
print("Testing Samples:", X_test.shape[0])
print("Accuracy:", round(accuracy * 100, 2), "%")
print("========================================")
```
