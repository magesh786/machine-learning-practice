```python
# ============================================================
# DAY 22 - ADVANCED LASSO REGRESSION
# California Housing Price Prediction
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import warnings
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import fetch_california_housing

from sklearn.model_selection import (
    train_test_split,
    KFold,
    GridSearchCV,
    learning_curve,
    cross_validate
)

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, PolynomialFeatures

from sklearn.linear_model import Lasso, LinearRegression

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

RANDOM_STATE = 42

OUTPUT_DIR = "day22_outputs"

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)


# ============================================================
# 2. LOAD DATASET
# ============================================================

housing = fetch_california_housing(
    as_frame=True
)

df = housing.frame.copy()

print("Dataset Shape:", df.shape)

display(df.head())


# ============================================================
# 3. BASIC DATA ANALYSIS
# ============================================================

print("\n========== DATASET INFORMATION ==========")

print(df.info())

print("\n========== MISSING VALUES ==========")

print(df.isnull().sum())

print("\n========== STATISTICAL SUMMARY ==========")

display(df.describe())


# ============================================================
# 4. FEATURE ENGINEERING
# ============================================================

df["Rooms_Per_Occupant"] = (
    df["AveRooms"] /
    df["AveOccup"].replace(0, np.nan)
)

df["Bedrooms_Per_Room"] = (
    df["AveBedrms"] /
    df["AveRooms"].replace(0, np.nan)
)

df["Population_Per_Room"] = (
    df["Population"] /
    df["AveRooms"].replace(0, np.nan)
)

df["Income_Age"] = (
    df["MedInc"] *
    df["HouseAge"]
)

df["Income_Rooms"] = (
    df["MedInc"] *
    df["AveRooms"]
)

df["Location_Interaction"] = (
    df["Latitude"] *
    df["Longitude"]
)

df = df.replace(
    [np.inf, -np.inf],
    np.nan
)


# ============================================================
# 5. FEATURES AND TARGET
# ============================================================

X = df.drop(
    "MedHouseVal",
    axis=1
)

y = df["MedHouseVal"]

print("\nNumber of Features:", X.shape[1])

print("\nFeatures:")

print(X.columns.tolist())


# ============================================================
# 6. TRAIN / VALIDATION / TEST SPLIT
# ============================================================

X_train_full, X_test, y_train_full, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE
)

X_train, X_validation, y_train, y_validation = train_test_split(
    X_train_full,
    y_train_full,
    test_size=0.20,
    random_state=RANDOM_STATE
)

print("\n========== DATA SPLIT ==========")

print("Training:", X_train.shape)

print("Validation:", X_validation.shape)

print("Testing:", X_test.shape)


# ============================================================
# 7. PREPROCESSING PIPELINE
# ============================================================

preprocessor = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),

        (
            "scaler",
            StandardScaler()
        )
    ]
)


# ============================================================
# 8. LASSO PIPELINE
# ============================================================

lasso_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),

        (
            "model",
            Lasso(
                max_iter=10000,
                random_state=RANDOM_STATE
            )
        )
    ]
)


# ============================================================
# 9. HYPERPARAMETER GRID
# ============================================================

alpha_values = np.logspace(
    -5,
    1,
    30
)

param_grid = {
    "model__alpha": alpha_values
}


# ============================================================
# 10. 5-FOLD CROSS VALIDATION
# ============================================================

cv = KFold(
    n_splits=5,
    shuffle=True,
    random_state=RANDOM_STATE
)


# ============================================================
# 11. GRID SEARCH
# ============================================================

grid_search = GridSearchCV(
    estimator=lasso_pipeline,
    param_grid=param_grid,
    scoring="neg_root_mean_squared_error",
    cv=cv,
    n_jobs=-1,
    return_train_score=True
)

print(
    "\n========== STARTING GRID SEARCH =========="
)

grid_search.fit(
    X_train,
    y_train
)

print(
    "\nGrid Search Completed!"
)


# ============================================================
# 12. BEST PARAMETER
# ============================================================

print(
    "\n========== BEST PARAMETER =========="
)

print(
    "Best Alpha:",
    grid_search.best_params_[
        "model__alpha"
    ]
)

print(
    "Best CV RMSE:",
    -grid_search.best_score_
)


# ============================================================
# 13. BEST LASSO MODEL
# ============================================================

best_model = grid_search.best_estimator_


# ============================================================
# 14. VALIDATION PREDICTION
# ============================================================

validation_predictions = best_model.predict(
    X_validation
)

validation_mae = mean_absolute_error(
    y_validation,
    validation_predictions
)

validation_mse = mean_squared_error(
    y_validation,
    validation_predictions
)

validation_rmse = np.sqrt(
    validation_mse
)

validation_r2 = r2_score(
    y_validation,
    validation_predictions
)

print(
    "\n========== VALIDATION PERFORMANCE =========="
)

print(
    f"MAE  : {validation_mae:.4f}"
)

print(
    f"MSE  : {validation_mse:.4f}"
)

print(
    f"RMSE : {validation_rmse:.4f}"
)

print(
    f"R²   : {validation_r2:.4f}"
)


# ============================================================
# 15. FINAL TEST PREDICTION
# ============================================================

test_predictions = best_model.predict(
    X_test
)

test_mae = mean_absolute_error(
    y_test,
    test_predictions
)

test_mse = mean_squared_error(
    y_test,
    test_predictions
)

test_rmse = np.sqrt(
    test_mse
)

test_r2 = r2_score(
    y_test,
    test_predictions
)

print(
    "\n========== FINAL TEST PERFORMANCE =========="
)

print(
    f"MAE  : {test_mae:.4f}"
)

print(
    f"MSE  : {test_mse:.4f}"
)

print(
    f"RMSE : {test_rmse:.4f}"
)

print(
    f"R²   : {test_r2:.4f}"
)


# ============================================================
# 16. LASSO COEFFICIENTS
# ============================================================

lasso_model = best_model.named_steps[
    "model"
]

coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": lasso_model.coef_
})

coefficients["Absolute_Coefficient"] = (
    coefficients["Coefficient"].abs()
)

coefficients["Selected"] = (
    coefficients["Coefficient"] != 0
)

coefficients = coefficients.sort_values(
    "Absolute_Coefficient",
    ascending=False
).reset_index(drop=True)

print(
    "\n========== LASSO COEFFICIENTS =========="
)

display(coefficients)


# ============================================================
# 17. FEATURE SELECTION ANALYSIS
# ============================================================

selected_features = coefficients[
    coefficients["Selected"]
]

removed_features = coefficients[
    ~coefficients["Selected"]
]

print(
    "\n========== FEATURE SELECTION =========="
)

print(
    "Total Features:",
    len(coefficients)
)

print(
    "Selected Features:",
    len(selected_features)
)

print(
    "Removed Features:",
    len(removed_features)
)

print(
    "\nSelected Features:"
)

print(
    selected_features["Feature"].tolist()
)

print(
    "\nFeatures Reduced to Zero:"
)

print(
    removed_features["Feature"].tolist()
)


# ============================================================
# 18. COEFFICIENT VISUALIZATION
# ============================================================

plot_data = coefficients.sort_values(
    "Coefficient"
)

plt.figure(
    figsize=(10, 7)
)

plt.barh(
    plot_data["Feature"],
    plot_data["Coefficient"]
)

plt.axvline(
    0,
    linestyle="--"
)

plt.xlabel(
    "Lasso Coefficient"
)

plt.ylabel(
    "Feature"
)

plt.title(
    "Lasso Regression Feature Coefficients"
)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/lasso_coefficients.png",
    dpi=150
)

plt.show()


# ============================================================
# 19. ACTUAL VS PREDICTED
# ============================================================

plt.figure(
    figsize=(9, 7)
)

plt.scatter(
    y_test,
    test_predictions,
    alpha=0.4
)

minimum = min(
    y_test.min(),
    test_predictions.min()
)

maximum = max(
    y_test.max(),
    test_predictions.max()
)

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--"
)

plt.xlabel(
    "Actual House Value"
)

plt.ylabel(
    "Predicted House Value"
)

plt.title(
    "Lasso Regression - Actual vs Predicted"
)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/actual_vs_predicted.png",
    dpi=150
)

plt.show()


# ============================================================
# 20. RESIDUAL ANALYSIS
# ============================================================

residuals = (
    y_test -
    test_predictions
)

plt.figure(
    figsize=(9, 6)
)

plt.scatter(
    test_predictions,
    residuals,
    alpha=0.4
)

plt.axhline(
    0,
    linestyle="--"
)

plt.xlabel(
    "Predicted House Value"
)

plt.ylabel(
    "Residual"
)

plt.title(
    "Lasso Regression Residual Analysis"
)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/residual_analysis.png",
    dpi=150
)

plt.show()


# ============================================================
# 21. RESIDUAL DISTRIBUTION
# ============================================================

plt.figure(
    figsize=(9, 6)
)

plt.hist(
    residuals,
    bins=40
)

plt.xlabel(
    "Residual"
)

plt.ylabel(
    "Frequency"
)

plt.title(
    "Residual Distribution"
)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/residual_distribution.png",
    dpi=150
)

plt.show()


# ============================================================
# 22. ALPHA EXPERIMENT
# ============================================================

alpha_experiment = []

for alpha in alpha_values:

    model = Pipeline(
        steps=[
            (
                "preprocessor",
                preprocessor
            ),

            (
                "model",
                Lasso(
                    alpha=alpha,
                    max_iter=10000,
                    random_state=RANDOM_STATE
                )
            )
        ]
    )

    model.fit(
        X_train,
        y_train
    )

    predictions = model.predict(
        X_validation
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_validation,
            predictions
        )
    )

    r2 = r2_score(
        y_validation,
        predictions
    )

    model_coefficients = (
        model.named_steps["model"].coef_
    )

    non_zero_features = np.sum(
        model_coefficients != 0
    )

    alpha_experiment.append({
        "Alpha": alpha,
        "RMSE": rmse,
        "R2": r2,
        "Non_Zero_Features":
            non_zero_features
    })


alpha_experiment = pd.DataFrame(
    alpha_experiment
)

display(
    alpha_experiment.head()
)


# ============================================================
# 23. ALPHA VS RMSE
# ============================================================

plt.figure(
    figsize=(9, 6)
)

plt.semilogx(
    alpha_experiment["Alpha"],
    alpha_experiment["RMSE"],
    marker="o"
)

plt.xlabel(
    "Alpha"
)

plt.ylabel(
    "Validation RMSE"
)

plt.title(
    "Lasso Alpha vs Validation RMSE"
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/alpha_vs_rmse.png",
    dpi=150
)

plt.show()


# ============================================================
# 24. ALPHA VS NUMBER OF FEATURES
# ============================================================

plt.figure(
    figsize=(9, 6)
)

plt.semilogx(
    alpha_experiment["Alpha"],
    alpha_experiment[
        "Non_Zero_Features"
    ],
    marker="o"
)

plt.xlabel(
    "Alpha"
)

plt.ylabel(
    "Number of Non-Zero Features"
)

plt.title(
    "Lasso Regularization vs Feature Selection"
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/alpha_vs_features.png",
    dpi=150
)

plt.show()


# ============================================================
# 25. CROSS-VALIDATION PERFORMANCE
# ============================================================

cv_results = cross_validate(
    best_model,
    X_train,
    y_train,
    cv=cv,
    scoring={
        "MAE":
            "neg_mean_absolute_error",

        "RMSE":
            "neg_root_mean_squared_error",

        "R2":
            "r2"
    },
    n_jobs=-1
)

cv_summary = pd.DataFrame({
    "Metric": [
        "MAE",
        "RMSE",
        "R2"
    ],

    "Mean": [
        -cv_results[
            "test_MAE"
        ].mean(),

        -cv_results[
            "test_RMSE"
        ].mean(),

        cv_results[
            "test_R2"
        ].mean()
    ],

    "Std": [
        cv_results[
            "test_MAE"
        ].std(),

        cv_results[
            "test_RMSE"
        ].std(),

        cv_results[
            "test_R2"
        ].std()
    ]
})

print(
    "\n========== CROSS VALIDATION =========="
)

display(
    cv_summary
)


# ============================================================
# 26. LEARNING CURVE
# ============================================================

train_sizes, train_scores, validation_scores = (
    learning_curve(
        best_model,
        X_train,
        y_train,
        cv=cv,
        scoring="neg_root_mean_squared_error",
        train_sizes=np.linspace(
            0.1,
            1.0,
            5
        ),
        n_jobs=-1
    )
)

train_rmse = -train_scores.mean(
    axis=1
)

validation_rmse = -validation_scores.mean(
    axis=1
)

plt.figure(
    figsize=(9, 6)
)

plt.plot(
    train_sizes,
    train_rmse,
    marker="o",
    label="Training RMSE"
)

plt.plot(
    train_sizes,
    validation_rmse,
    marker="o",
    label="Validation RMSE"
)

plt.xlabel(
    "Training Samples"
)

plt.ylabel(
    "RMSE"
)

plt.title(
    "Lasso Regression Learning Curve"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/learning_curve.png",
    dpi=150
)

plt.show()


# ============================================================
# 27. PREDICTION RESULTS
# ============================================================

prediction_results = X_test.copy()

prediction_results[
    "Actual_House_Value"
] = y_test.values

prediction_results[
    "Predicted_House_Value"
] = test_predictions

prediction_results[
    "Residual"
] = residuals.values

prediction_results[
    "Absolute_Error"
] = np.abs(
    residuals.values
)

display(
    prediction_results.head()
)


# ============================================================
# 28. LARGEST ERRORS
# ============================================================

largest_errors = (
    prediction_results
    .sort_values(
        "Absolute_Error",
        ascending=False
    )
    .head(20)
)

print(
    "\n========== LARGEST ERRORS =========="
)

display(
    largest_errors
)


# ============================================================
# 29. SAVE RESULTS
# ============================================================

prediction_results.to_csv(
    f"{OUTPUT_DIR}/lasso_house_predictions.csv",
    index=False
)

coefficients.to_csv(
    f"{OUTPUT_DIR}/lasso_coefficients.csv",
    index=False
)

alpha_experiment.to_csv(
    f"{OUTPUT_DIR}/lasso_alpha_experiment.csv",
    index=False
)

cv_summary.to_csv(
    f"{OUTPUT_DIR}/lasso_cross_validation.csv",
    index=False
)

pd.DataFrame(
    grid_search.cv_results_
).to_csv(
    f"{OUTPUT_DIR}/lasso_grid_search.csv",
    index=False
)


# ============================================================
# 30. MODEL SUMMARY
# ============================================================

model_summary = pd.DataFrame({
    "Metric": [
        "Algorithm",
        "Number of Features",
        "Selected Features",
        "Removed Features",
        "Best Alpha",
        "Best CV RMSE",
        "Validation MAE",
        "Validation RMSE",
        "Validation R2",
        "Test MAE",
        "Test RMSE",
        "Test R2"
    ],

    "Value": [
        "Lasso Regression",
        len(coefficients),
        len(selected_features),
        len(removed_features),
        grid_search.best_params_[
            "model__alpha"
        ],
        -grid_search.best_score_,
        validation_mae,
        validation_rmse,
        validation_r2,
        test_mae,
        test_rmse,
        test_r2
    ]
})

model_summary.to_csv(
    f"{OUTPUT_DIR}/lasso_model_summary.csv",
    index=False
)


# ============================================================
# 31. SAVE MODEL
# ============================================================

joblib.dump(
    best_model,
    f"{OUTPUT_DIR}/lasso_house_model.pkl"
)

print(
    "\nModel saved successfully!"
)


# ============================================================
# 32. CUSTOM PREDICTION FUNCTION
# ============================================================

def predict_house(
    MedInc,
    HouseAge,
    AveRooms,
    AveBedrms,
    Population,
    AveOccup,
    Latitude,
    Longitude
):

    new_house = pd.DataFrame([{
        "MedInc": MedInc,
        "HouseAge": HouseAge,
        "AveRooms": AveRooms,
        "AveBedrms": AveBedrms,
        "Population": Population,
        "AveOccup": AveOccup,
        "Latitude": Latitude,
        "Longitude": Longitude
    }])

    new_house[
        "Rooms_Per_Occupant"
    ] = (
        new_house["AveRooms"] /
        new_house["AveOccup"].replace(
            0,
            np.nan
        )
    )

    new_house[
        "Bedrooms_Per_Room"
    ] = (
        new_house["AveBedrms"] /
        new_house["AveRooms"].replace(
            0,
            np.nan
        )
    )

    new_house[
        "Population_Per_Room"
    ] = (
        new_house["Population"] /
        new_house["AveRooms"].replace(
            0,
            np.nan
        )
    )

    new_house[
        "Income_Age"
    ] = (
        new_house["MedInc"] *
        new_house["HouseAge"]
    )

    new_house[
        "Income_Rooms"
    ] = (
        new_house["MedInc"] *
        new_house["AveRooms"]
    )

    new_house[
        "Location_Interaction"
    ] = (
        new_house["Latitude"] *
        new_house["Longitude"]
    )

    new_house = new_house.replace(
        [np.inf, -np.inf],
        np.nan
    )

    prediction = best_model.predict(
        new_house
    )[0]

    return prediction


# ============================================================
# 33. CUSTOM PREDICTION
# ============================================================

prediction = predict_house(
    MedInc=5.0,
    HouseAge=20,
    AveRooms=6.0,
    AveBedrms=1.0,
    Population=1000,
    AveOccup=3.0,
    Latitude=34.0,
    Longitude=-118.0
)

print(
    "\n========== CUSTOM PREDICTION =========="
)

print(
    f"Predicted House Value: "
    f"${prediction * 100000:,.2f}"
)


# ============================================================
# 34. FINAL SUMMARY
# ============================================================

print("\n==========================================")
print("DAY 22 COMPLETED")
print("ADVANCED LASSO REGRESSION")
print("==========================================")

print(
    f"Best Alpha : "
    f"{grid_search.best_params_['model__alpha']}"
)

print(
    f"Test RMSE  : {test_rmse:.4f}"
)

print(
    f"Test R²    : {test_r2:.4f}"
)

print(
    f"Selected Features : "
    f"{len(selected_features)}"
)

print(
    f"Removed Features  : "
    f"{len(removed_features)}"
)

print("\nGenerated files:")

for file in os.listdir(
    OUTPUT_DIR
):
    print(
        "-",
        file
    )
```
