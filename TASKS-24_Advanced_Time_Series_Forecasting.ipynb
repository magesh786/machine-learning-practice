```python
# ============================================================
# DAY 24 - ADVANCED TIME SERIES FORECASTING
# Electricity Demand Forecasting using Random Forest
# ============================================================

# =========================
# 1. IMPORT LIBRARIES
# =========================

import os
import warnings
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import TimeSeriesSplit, RandomizedSearchCV

warnings.filterwarnings("ignore")

OUTPUT_DIR = "day24_outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

RANDOM_STATE = 42

print("Libraries imported successfully.")


# =========================
# 2. CREATE SYNTHETIC HOURLY DATA
# =========================

np.random.seed(RANDOM_STATE)

date_range = pd.date_range(
    start="2024-01-01",
    end="2025-12-31 23:00:00",
    freq="h"
)

df = pd.DataFrame({"timestamp": date_range})

# Time components
df["hour"] = df["timestamp"].dt.hour
df["day_of_week"] = df["timestamp"].dt.dayofweek
df["day_of_year"] = df["timestamp"].dt.dayofyear
df["month"] = df["timestamp"].dt.month

# Weekend indicator
df["is_weekend"] = (df["day_of_week"] >= 5).astype(int)

# Seasonal components
daily_pattern = (
    300
    + 120 * np.sin(2 * np.pi * (df["hour"] - 7) / 24)
)

weekly_pattern = (
    -50 * np.cos(2 * np.pi * df["day_of_week"] / 7)
)

yearly_pattern = (
    100 * np.sin(2 * np.pi * df["day_of_year"] / 365.25)
)

# Temperature-like seasonal signal
temperature = (
    27
    + 7 * np.sin(2 * np.pi * (df["day_of_year"] - 80) / 365.25)
    + 3 * np.sin(2 * np.pi * df["hour"] / 24)
    + np.random.normal(0, 1.5, len(df))
)

df["temperature"] = temperature

# Electricity demand
df["demand"] = (
    900
    + daily_pattern
    + weekly_pattern
    + yearly_pattern
    + 18 * np.maximum(df["temperature"] - 28, 0)
    + 35 * df["is_weekend"]
    + np.random.normal(0, 35, len(df))
)

df["demand"] = df["demand"].clip(lower=400)

print("Dataset shape:", df.shape)
display(df.head())


# =========================
# 3. CHECK DATA
# =========================

print("\nDataset Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
display(df.describe())


# =========================
# 4. VISUALIZE ORIGINAL SERIES
# =========================

plt.figure(figsize=(16, 5))
plt.plot(df["timestamp"], df["demand"])
plt.title("Hourly Electricity Demand")
plt.xlabel("Time")
plt.ylabel("Demand")
plt.grid(True)
plt.tight_layout()
plt.savefig(
    f"{OUTPUT_DIR}/original_demand.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()


# =========================
# 5. FEATURE ENGINEERING
# =========================

# Lag features
df["lag_1"] = df["demand"].shift(1)
df["lag_2"] = df["demand"].shift(2)
df["lag_3"] = df["demand"].shift(3)
df["lag_24"] = df["demand"].shift(24)
df["lag_48"] = df["demand"].shift(48)
df["lag_168"] = df["demand"].shift(168)

# Rolling statistics
df["rolling_mean_24"] = (
    df["demand"]
    .shift(1)
    .rolling(window=24)
    .mean()
)

df["rolling_std_24"] = (
    df["demand"]
    .shift(1)
    .rolling(window=24)
    .std()
)

df["rolling_mean_168"] = (
    df["demand"]
    .shift(1)
    .rolling(window=168)
    .mean()
)

df["rolling_std_168"] = (
    df["demand"]
    .shift(1)
    .rolling(window=168)
    .std()
)

# Cyclic time encoding
df["hour_sin"] = np.sin(2 * np.pi * df["hour"] / 24)
df["hour_cos"] = np.cos(2 * np.pi * df["hour"] / 24)

df["day_sin"] = np.sin(2 * np.pi * df["day_of_week"] / 7)
df["day_cos"] = np.cos(2 * np.pi * df["day_of_week"] / 7)

df["month_sin"] = np.sin(2 * np.pi * df["month"] / 12)
df["month_cos"] = np.cos(2 * np.pi * df["month"] / 12)

# Remove rows created by lagging
df_model = df.dropna().reset_index(drop=True)

print("Feature-engineered dataset:", df_model.shape)

display(df_model.head())


# =========================
# 6. DEFINE FEATURES
# =========================

features = [
    "hour",
    "day_of_week",
    "day_of_year",
    "month",
    "is_weekend",
    "temperature",

    "lag_1",
    "lag_2",
    "lag_3",
    "lag_24",
    "lag_48",
    "lag_168",

    "rolling_mean_24",
    "rolling_std_24",
    "rolling_mean_168",
    "rolling_std_168",

    "hour_sin",
    "hour_cos",
    "day_sin",
    "day_cos",
    "month_sin",
    "month_cos"
]

X = df_model[features]
y = df_model["demand"]


# =========================
# 7. TIME-BASED TRAIN/TEST SPLIT
# =========================

# IMPORTANT:
# Do not randomly shuffle time-series data.

split_index = int(len(df_model) * 0.80)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

test_timestamps = df_model["timestamp"].iloc[split_index:]

print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))

print("\nTraining period:")
print(df_model["timestamp"].iloc[0])
print(df_model["timestamp"].iloc[split_index - 1])

print("\nTesting period:")
print(df_model["timestamp"].iloc[split_index])
print(df_model["timestamp"].iloc[-1])


# =========================
# 8. BASE RANDOM FOREST MODEL
# =========================

base_model = RandomForestRegressor(
    n_estimators=300,
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    max_features="sqrt",
    random_state=RANDOM_STATE,
    n_jobs=-1
)

base_model.fit(X_train, y_train)

base_pred = base_model.predict(X_test)

base_mae = mean_absolute_error(y_test, base_pred)
base_rmse = np.sqrt(mean_squared_error(y_test, base_pred))
base_r2 = r2_score(y_test, base_pred)

print("\nBase Model Performance")
print("-" * 40)
print(f"MAE  : {base_mae:.4f}")
print(f"RMSE : {base_rmse:.4f}")
print(f"R²   : {base_r2:.4f}")


# =========================
# 9. HYPERPARAMETER TUNING
# =========================

param_distributions = {
    "n_estimators": [200, 300, 500, 700],
    "max_depth": [None, 10, 15, 20, 25, 30],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4],
    "max_features": ["sqrt", "log2", 0.5, 0.8]
}

tscv = TimeSeriesSplit(n_splits=5)

random_search = RandomizedSearchCV(
    estimator=RandomForestRegressor(
        random_state=RANDOM_STATE,
        n_jobs=-1
    ),
    param_distributions=param_distributions,
    n_iter=20,
    scoring="neg_root_mean_squared_error",
    cv=tscv,
    random_state=RANDOM_STATE,
    n_jobs=-1,
    verbose=1
)

random_search.fit(X_train, y_train)

best_model = random_search.best_estimator_

print("\nBest Parameters:")
print(random_search.best_params_)

print("\nBest CV RMSE:")
print(-random_search.best_score_)


# =========================
# 10. SAVE HYPERPARAMETER RESULTS
# =========================

search_results = pd.DataFrame(random_search.cv_results_)

search_results = search_results.sort_values(
    by="rank_test_score"
)

search_results.to_csv(
    f"{OUTPUT_DIR}/hyperparameter_search.csv",
    index=False
)

pd.DataFrame(
    [random_search.best_params_]
).to_csv(
    f"{OUTPUT_DIR}/best_parameters.csv",
    index=False
)


# =========================
# 11. FINAL TEST PREDICTIONS
# =========================

y_pred = best_model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("\nFinal Test Performance")
print("=" * 45)
print(f"MAE  : {mae:.4f}")
print(f"MSE  : {mse:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"R²   : {r2:.4f}")


# =========================
# 12. SAVE MODEL SUMMARY
# =========================

metrics_df = pd.DataFrame({
    "Metric": [
        "MAE",
        "MSE",
        "RMSE",
        "R2"
    ],
    "Value": [
        mae,
        mse,
        rmse,
        r2
    ]
})

metrics_df.to_csv(
    f"{OUTPUT_DIR}/model_metrics.csv",
    index=False
)


# =========================
# 13. PREDICTION DATAFRAME
# =========================

prediction_df = pd.DataFrame({
    "timestamp": test_timestamps.values,
    "actual_demand": y_test.values,
    "predicted_demand": y_pred
})

prediction_df["error"] = (
    prediction_df["actual_demand"]
    - prediction_df["predicted_demand"]
)

prediction_df["absolute_error"] = (
    prediction_df["error"].abs()
)

prediction_df.to_csv(
    f"{OUTPUT_DIR}/forecast_predictions.csv",
    index=False
)

display(prediction_df.head(20))


# =========================
# 14. ACTUAL VS PREDICTED
# =========================

plt.figure(figsize=(16, 6))

plt.plot(
    prediction_df["timestamp"],
    prediction_df["actual_demand"],
    label="Actual"
)

plt.plot(
    prediction_df["timestamp"],
    prediction_df["predicted_demand"],
    label="Predicted"
)

plt.title("Actual vs Predicted Electricity Demand")
plt.xlabel("Time")
plt.ylabel("Demand")
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/actual_vs_predicted.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =========================
# 15. LAST 7 DAYS FORECAST
# =========================

last_hours = 24 * 7

recent_predictions = prediction_df.tail(last_hours)

plt.figure(figsize=(16, 6))

plt.plot(
    recent_predictions["timestamp"],
    recent_predictions["actual_demand"],
    marker="o",
    markersize=2,
    label="Actual"
)

plt.plot(
    recent_predictions["timestamp"],
    recent_predictions["predicted_demand"],
    marker="o",
    markersize=2,
    label="Predicted"
)

plt.title("Last 7 Days: Actual vs Predicted Demand")
plt.xlabel("Time")
plt.ylabel("Demand")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/last_7_days_forecast.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =========================
# 16. RESIDUAL ANALYSIS
# =========================

residuals = y_test.values - y_pred

plt.figure(figsize=(12, 5))

plt.plot(
    test_timestamps,
    residuals
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.title("Forecast Residuals Over Time")
plt.xlabel("Time")
plt.ylabel("Residual")
plt.grid(True)
plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/residual_analysis.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =========================
# 17. RESIDUAL DISTRIBUTION
# =========================

plt.figure(figsize=(10, 5))

plt.hist(
    residuals,
    bins=50
)

plt.title("Residual Distribution")
plt.xlabel("Residual")
plt.ylabel("Frequency")
plt.grid(True)
plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/residual_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =========================
# 18. FEATURE IMPORTANCE
# =========================

importance_df = pd.DataFrame({
    "feature": features,
    "importance": best_model.feature_importances_
})

importance_df = importance_df.sort_values(
    by="importance",
    ascending=False
)

importance_df.to_csv(
    f"{OUTPUT_DIR}/feature_importance.csv",
    index=False
)

display(importance_df)


plt.figure(figsize=(12, 8))

top_features = importance_df.head(15)

plt.barh(
    top_features["feature"][::-1],
    top_features["importance"][::-1]
)

plt.title("Top 15 Feature Importances")
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/feature_importance.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =========================
# 19. WALK-FORWARD VALIDATION
# =========================

walk_forward_results = []

fold_number = 1

for train_index, validation_index in tscv.split(X_train):

    X_tr = X_train.iloc[train_index]
    X_val = X_train.iloc[validation_index]

    y_tr = y_train.iloc[train_index]
    y_val = y_train.iloc[validation_index]

    fold_model = RandomForestRegressor(
        n_estimators=300,
        max_features="sqrt",
        random_state=RANDOM_STATE,
        n_jobs=-1
    )

    fold_model.fit(X_tr, y_tr)

    fold_pred = fold_model.predict(X_val)

    fold_mae = mean_absolute_error(y_val, fold_pred)
    fold_rmse = np.sqrt(
        mean_squared_error(y_val, fold_pred)
    )
    fold_r2 = r2_score(y_val, fold_pred)

    walk_forward_results.append({
        "fold": fold_number,
        "MAE": fold_mae,
        "RMSE": fold_rmse,
        "R2": fold_r2
    })

    fold_number += 1


walk_forward_df = pd.DataFrame(walk_forward_results)

display(walk_forward_df)

walk_forward_df.to_csv(
    f"{OUTPUT_DIR}/walk_forward_validation.csv",
    index=False
)


# =========================
# 20. MODEL COMPARISON
# =========================

comparison_df = pd.DataFrame({
    "Model": [
        "Base Random Forest",
        "Tuned Random Forest"
    ],
    "MAE": [
        base_mae,
        mae
    ],
    "RMSE": [
        base_rmse,
        rmse
    ],
    "R2": [
        base_r2,
        r2
    ]
})

display(comparison_df)

comparison_df.to_csv(
    f"{OUTPUT_DIR}/model_comparison.csv",
    index=False
)


# =========================
# 21. LARGEST FORECAST ERRORS
# =========================

largest_errors = prediction_df.sort_values(
    by="absolute_error",
    ascending=False
).head(20)

display(largest_errors)

largest_errors.to_csv(
    f"{OUTPUT_DIR}/largest_forecast_errors.csv",
    index=False
)


# =========================
# 22. SAVE TRAINED MODEL
# =========================

model_path = (
    f"{OUTPUT_DIR}/electricity_demand_forecasting_model.pkl"
)

joblib.dump(
    best_model,
    model_path
)

print(f"\nModel saved to: {model_path}")


# =========================
# 23. LOAD MODEL
# =========================

loaded_model = joblib.load(model_path)

loaded_predictions = loaded_model.predict(
    X_test.head(10)
)

print("\nSample predictions from saved model:")
print(loaded_predictions)


# =========================
# 24. CUSTOM FUTURE-LIKE PREDICTION
# =========================

sample_input = X_test.iloc[[0]]

sample_prediction = loaded_model.predict(
    sample_input
)[0]

print("\nSample Electricity Demand Prediction:")
print(f"{sample_prediction:.2f}")


# =========================
# 25. FINAL SUMMARY
# =========================

print("\n" + "=" * 60)
print("DAY 24 - TIME SERIES FORECASTING COMPLETED")
print("=" * 60)

print(f"MAE  : {mae:.4f}")
print(f"MSE  : {mse:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"R²   : {r2:.4f}")

print("\nTop 10 Features:")
display(importance_df.head(10))

print("\nOutput files:")
for file in sorted(os.listdir(OUTPUT_DIR)):
    print("-", file)
```
